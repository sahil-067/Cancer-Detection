This notebook demonstrates how to run inference using the trained YOLO model on test images.

```python
import os
from ultralytics import YOLO
import matplotlib.pyplot as plt
import cv2
import glob

# Check if CUDA is available
import torch
device = 'cuda' if torch.cuda.is_available() else 'cpu'
print(f'Using device: {device}')
```

```python
# Load the model
model_path = 'Model/best.pt'
if not os.path.exists(model_path):
    print(f"Error: Model not found at {model_path}")
else:
    model = YOLO(model_path)
    print("Model loaded successfully")
```

```python
# Path to test images
test_images_path = 'test_data/images'

# Get list of images (supporting jpg, png, jpeg)
image_extensions = ['*.jpg', '*.jpeg', '*.png']
images = []
for ext in image_extensions:
    images.extend(glob.glob(os.path.join(test_images_path, ext)))

print(f"Found {len(images)} images in {test_images_path}")
```

```python
# Run inference on a few random sample images
import random
num_samples = 5

# Select random images if there are enough images
if len(images) > num_samples:
    selected_images = random.sample(images, num_samples)
else:
    selected_images = images

for img_path in selected_images:
    print(f"Running inference on: {os.path.basename(img_path)}")
    results = model(img_path)
    
    # Visualize results
    for result in results:
        im_array = result.plot()  # plot a BGR numpy array of predictions
        im_rgb = cv2.cvtColor(im_array, cv2.COLOR_BGR2RGB)
        
        plt.figure(figsize=(10, 10))
        plt.imshow(im_rgb)
        plt.axis('off')
        plt.title(f"Prediction: {os.path.basename(img_path)}")
        plt.show()
```
